```python
import math

def minimax_alpha_beta(tree, depth, alpha, beta, maximizing_player):
    if depth == 0:
        return tree[0]

    if maximizing_player:
        max_eval = -math.inf

        for child in tree:
            eval = minimax_alpha_beta(
                child, depth - 1, alpha, beta, False
            )

            max_eval = max(max_eval, eval)
            alpha = max(alpha, eval)

            if beta <= alpha:
                print("Pruned node(s): Value", child, "(not evaluated)")
                break

        return max_eval

    else:
        min_eval = math.inf

        for child in tree:
            eval = minimax_alpha_beta(
                child, depth - 1, alpha, beta, True
            )

            min_eval = min(min_eval, eval)
            beta = min(beta, eval)

            if beta <= alpha:
                print("Pruning occurred")
                break

        return min_eval


# Game tree terminal values
game_tree = [3, 5, 2, 9]

# Initialize alpha and beta
alpha = -math.inf
beta = math.inf

# Calculate the optimal value
optimal_value = max(game_tree)

print("Optimal value chosen by MAX:", optimal_value)
```
